```python
import numpy as np
from pathlib import Path
from scipy.io import wavfile

# -------------------------
# RET DENNE STI
# -------------------------
IN_DIR = Path(r"unknown")      # <-- ret hvis nødvendigt
OUT_DIR = IN_DIR / "spliced"

CHUNK_DURATION = 1.0  # sekunder

# -------------------------

def slice_unknown_file(wav_path: Path):
    fs, data = wavfile.read(wav_path)

    # mono
    if data.ndim == 2:
        data = data.mean(axis=1)

    data = data.astype(np.int16)

    samples_per_chunk = int(fs * CHUNK_DURATION)
    total_samples = len(data)

    n_chunks = total_samples // samples_per_chunk

    OUT_DIR.mkdir(parents=True, exist_ok=True)

    written = 0

    for i in range(n_chunks):
        start = i * samples_per_chunk
        end = start + samples_per_chunk

        chunk = data[start:end]

        out_name = f"{wav_path.stem}_u{i+1:02d}.wav"
        out_path = OUT_DIR / out_name

        wavfile.write(out_path, fs, chunk)
        written += 1

    return written


def main():
    if not IN_DIR.exists():
        raise SystemExit(f"Folder findes ikke: {IN_DIR.resolve()}")

    wav_files = sorted(IN_DIR.glob("*.wav"))

    print(f"Fandt {len(wav_files)} unknown-filer")

    total = 0
    for wav in wav_files:
        n = slice_unknown_file(wav)
        print(f"{wav.name}: {n} chunks")
        total += n

    print("Total unknown slices:", total)
    print("Output:", OUT_DIR.resolve())


if __name__ == "__main__":
    main()
```
