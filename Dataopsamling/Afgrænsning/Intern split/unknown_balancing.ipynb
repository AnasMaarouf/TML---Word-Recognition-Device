```python
from pathlib import Path
import shutil
import random

ROOT = Path(".").resolve()
UNKNOWN_DIR = ROOT / "unknown"
OUT_DIR = ROOT / "unknown_balanced"

# ønsket fordeling
TARGET_COUNTS = {
    "random": 300,
    "stilhed": 200,
    #"storcenter": 150,
    "nygaard": 200,
    "tmlKlasse": 200,
}

def collect_wavs(folder: Path):
    return sorted([p for p in folder.glob("*.wav") if p.is_file()])

def ensure_clean_dir(path: Path):
    if path.exists():
        shutil.rmtree(path)
    path.mkdir(parents=True, exist_ok=True)

def main():
    random.seed(42)
    ensure_clean_dir(OUT_DIR)

    total = 0

    for label, target_n in TARGET_COUNTS.items():
        src_dir = UNKNOWN_DIR / label / "spliced"
        files = collect_wavs(src_dir)

        if not files:
            print(f"[WARN] No wav files found in {src_dir}")
            continue

        print(f"[{label}] available: {len(files)}, target: {target_n}")

        if len(files) >= target_n:
            chosen = random.sample(files, target_n)
        else:
            chosen = files.copy()
            extra = random.choices(files, k=target_n - len(files))
            chosen.extend(extra)

        for i, src in enumerate(chosen, start=1):
            out_name = f"unknown_{label}_{i:04d}.wav"
            shutil.copy2(src, OUT_DIR / out_name)

        total += len(chosen)

    print(f"\nDone. Wrote {total} unknown samples to {OUT_DIR}")

if __name__ == "__main__":
    main()
```
